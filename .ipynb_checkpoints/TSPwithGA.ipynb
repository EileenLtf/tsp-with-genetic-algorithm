```python
import numpy as np 
```

```python
def initializing_the_graph():
    distances = np.zeros((10, 10))
    lower_triangle = np.tril_indices(10, k=-1)
    distances[lower_triangle] = np.random.randint(10, 101, size=len(lower_triangle[0]))
    return distances
```

```python
initializing_the_graph()
```

```python
def chromosome_maker():
    array = np.random.choice(np.arange(1, 10), size = 9, replace = False) 
    start_city = np.array([0])
    array = np.concatenate((start_city, array))
    return array 
    
```

```python
chromosome_maker()
```

```python
def Population_maker_using_uniform_random(N): 
    population = set()
    while len(population) < N:
        newchromosome = tuple(chromosome_maker())
        population.add(newchromosome)
    return np.array(list(population))
```

```python
population = Population_maker_using_uniform_random(50)
```

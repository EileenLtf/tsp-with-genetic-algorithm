```python
import numpy as np 
np.random.seed(100)
```

```python
def initializing_the_graph():
    distances = np.zeros((10, 10))
    lower_triangle = np.tril_indices(10, k=-1)
    distances[lower_triangle] = np.random.randint(10, 101, size=len(lower_triangle[0]))
    distances += distances.T
    return distances
```

```python
 distances = initializing_the_graph()
distances 
```

```python
def chromosome_maker():
    array = np.random.choice(np.arange(1, 10), size = 9, replace = False) 
    start_city = np.array([0])
    array = np.concatenate((start_city, array))
    return array 
    
```

```python
chromosome_maker()
```

```python
def Population_maker_using_uniform_random(N): 
    population = set()
    while len(population) < N:
        newchromosome = tuple(chromosome_maker())
        population.add(newchromosome)
    return np.array(list(population))
```

```python
population = Population_maker_using_uniform_random(50)
```

```python
def FitessFunction(population, distances):
    zero_coulmn = np.zeros((population.shape[0], 1), dtype=int)
    supplementary_graph = np.hstack([population[:, 1:], zero_coulmn])
    all_distances = distances[population, supplementary_graph].sum(axis=1)
    F_of_x = 1000/(all_distances)
    return F_of_x
```

```python
Fitness = FitessFunction(population, distances)
Fitness
```

```python
def Selection_with_tounament(k, Fitness, population):
    tournamentors = np.random.choice(range(0, 50), size=k, replace=False)
    # print(tournamentors)
    # winner = np.max(Fitness[tournamentors])
    winner_index = np.argmax(Fitness[tournamentors])
    what = tournamentors[winner_index]
    # print(what)
    return population[what]
```

```python
winner1 = Selection_with_tounament(3, Fitness, population)
winner2 = Selection_with_tounament(3, Fitness, population)
```

```python
def ox_crossover(parent1, parent2):
    a, b =np.sort(np.random.choice(len(parent1), size=2, replace=False))
    offspring = np.full(len(parent1), -1, dtype=int)
    # offspring[0] = 0 
    # print(a, b)
    offspring[a:b] = parent1[a:b]
    temp = np.isin(parent2, offspring, invert=True)
    # print(temp)
    # print(offspring)
    availabe_in_parent2 = parent2[temp]
    # print(availabe_in_parent2)
    offspring[offspring == -1 ] = availabe_in_parent2
    # print(offspring)
    return offspring 
```

```python
offspring = ox_crossover(winner1, winner2) 
```

```python
def mutation_with_inversion(rate, chromosome):
    dice = np.random.rand()
    if dice <= rate: 
        a, b =np.sort(np.random.choice(len(chromosome), size=2, replace=False))
        # print(a, b)
        reversed = np.flip(chromosome[a : b + 1])
        # print(reversed)
        chromosome[a:b+1 ] = reversed 
    return chromosome 
```
